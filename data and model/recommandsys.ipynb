```python
import ast
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
import ast

def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:   
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
import ast

def directorcrew(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'].lower() == 'director':   # safer: case-insensitive
            L.append(i['name'])
            break   # stop after first director found
    return L
```

```python
movies['crew']=movies['crew'].apply(directorcrew)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(vectors)
```

```python
cosine_similarity(vectors).shape
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
similarity[10]
```

```python
def recommend(movie):
    # find the index of the movie
    movie_index = n_df[n_df['title'] == movie].index[0]
    
    # get similarity scores for that movie
    distances = similarity[movie_index]
    
    # sort by similarity (excluding the movie itself)
    movie_list = sorted(list(enumerate(distances)), 
                        key=lambda x: x[1], 
                        reverse=True)[1:6]
    
    # return the top 5 similar movies
    return [n_df.iloc[i[0]].title for i in movie_list]
```

```python
print(recommend("Avatar"))
```

```python
print(recommend("Batman Begins"))
```

```python
print(recommend("Titanic"))
```
